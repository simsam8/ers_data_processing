```python
import pandas as pd
import os
```

```python
ais_frames = []
for ais_file in os.listdir("processed/ais/AIS2018/"):
    temp = pd.read_parquet("processed/ais/AIS2018/" + ais_file)
    ais_frames.append(temp)
    
for ais_file in os.listdir("processed/ais/AIS2019/"):
    temp = pd.read_parquet("processed/ais/AIS2019/" + ais_file)
    ais_frames.append(temp)
    
for ais_file in os.listdir("processed/ais/AIS2020/"):
    temp = pd.read_parquet("processed/ais/AIS2020/" + ais_file)
    ais_frames.append(temp)

ais_frame = pd.concat(ais_frames)
```
